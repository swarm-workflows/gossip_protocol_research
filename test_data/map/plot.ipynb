```python
# Make sure you have Cartopy installed:
#    pip install cartopy
# Also ensure you have matplotlib installed:
#    pip install matplotlib

import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# Approximate lat/long coordinates for each site (modify as needed)
site_coords = {
    "SRI":    (37.4580, -122.1700),  # SRI (Menlo Park, CA)
    "MICH":   (42.2808,  -83.7430),  # Michigan (Ann Arbor, MI)
    "BRIST":  (51.4545,   -2.5879),  # Bristol (UK)
    "MAX":    (38.9881,  -77.0940),  # Mid-Atlantic Crossroads/DC area
    "CERN":   (46.2330,    6.0550),  # Near Geneva, Switzerland
    "UCSD":   (32.8801, -117.2340),  # San Diego, CA
    "AMST":   (52.3676,    4.9041),  # Amsterdam (Netherlands)
    "GATECH": (33.7756,  -84.3963),  # Atlanta, GA
    "MASS":   (42.3601,  -71.0589),  # Massachusetts (Boston area)
    "UTAH":   (40.7608, -111.8910),  # Salt Lake City, UT
    "KANS":   (39.0119,  -98.4842),  # Kansas (central approx)
    # "GPN":    (39.0997,  -94.5786),  # Great Plains Net (Kansas City, MO)
    "FIU":    (25.7565,  -80.3760),  # Miami, FL
    "PRIN":   (40.3573,  -74.6672),  # Princeton, NJ
    "HAWI":   (21.3069, -157.8583),  # Hawaii (Honolulu)
    "LOSA":   (34.0522, -118.2437),  # Los Angeles, CA
    "EDC":    (38.9072,  -77.0369),  # EDC (Washington, DC area)
    # "TACC":   (30.2849,  -97.7341),  # Austin, TX
    "INDI":   (39.1653,  -86.5264),  # Bloomington, IN
}

def plot_sites_and_save_pdf(site_coords, output_pdf="fabric_sites.pdf"):
    """
    Plot approximate site locations on a world map and save as a PDF.
    """

    # Create a figure and axis with Cartopy's PlateCarree (lat/lon) projection
    fig = plt.figure(figsize=(12, 6))
    ax = fig.add_subplot(1, 1, 1, projection=ccrs.PlateCarree())

    # Optionally set global bounds or extent
    ax.set_global()

    # Add map features (land, ocean, borders, etc.)
    ax.add_feature(cfeature.LAND, edgecolor='black', zorder=0)
    ax.add_feature(cfeature.OCEAN, zorder=0)
    ax.add_feature(cfeature.BORDERS, linestyle=':', zorder=1)
    ax.add_feature(cfeature.LAKES, alpha=0.5, zorder=1)
    ax.add_feature(cfeature.RIVERS, zorder=1)

    # Plot each site on the map
    for site_name, (lat, lon) in site_coords.items():
        ax.plot(lon, lat,
                marker='o',
                color='red',
                markersize=5,
                transform=ccrs.PlateCarree())
        # Label each site slightly offset from its marker
        ax.text(lon + 1.0, lat + 0.5, site_name,
                transform=ccrs.PlateCarree(),
                fontsize=9,
                color='blue')

    # Save to PDF
    plt.savefig(output_pdf, format='pdf', dpi=300)
    plt.show()

# Call the function to generate and save the map
plot_sites_and_save_pdf(site_coords, "fabric_sites.pdf")
```

```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature

# 在这里列出你想要显示的站点及其大致坐标 (纬度, 经度)
site_coords = {
    "SRI":    (37.4580, -122.1700),  # SRI (Menlo Park, CA)
    "MICH":   (42.2808,  -83.7430),  # Michigan (Ann Arbor, MI)
    "BRIST":  (51.4545,   -2.5879),  # Bristol (UK)
    "MAX":    (38.9881,  -77.0940),  # Mid-Atlantic Crossroads/DC area
    "CERN":   (46.2330,    6.0550),  # Near Geneva, Switzerland
    "UCSD":   (32.8801, -117.2340),  # San Diego, CA
    "AMST":   (52.3676,    4.9041),  # Amsterdam (Netherlands)
    "GATECH": (33.7756,  -84.3963),  # Atlanta, GA
    "MASS":   (42.3601,  -71.0589),  # Massachusetts (Boston area)
    "UTAH":   (40.7608, -111.8910),  # Salt Lake City, UT
    "KANS":   (39.0119,  -98.4842),  # Kansas (central approx)
    # "GPN":    (39.0997,  -94.5786),  # Great Plains Net (Kansas City, MO)
    "FIU":    (25.7565,  -80.3760),  # Miami, FL
    "PRIN":   (40.3573,  -74.6672),  # Princeton, NJ
    "HAWI":   (21.3069, -157.8583),  # Hawaii (Honolulu)
    "LOSA":   (34.0522, -118.2437),  # Los Angeles, CA
    "EDC":    (38.9072,  -77.0369),  # EDC (Washington, DC area)
    "TACC":   (30.2849,  -97.7341),  # Austin, TX
    "INDI":   (39.1653,  -86.5264),  # Bloomington, IN
}

def plot_sites_zoom_in(site_coords, output_pdf="fabric_sites_zoom.pdf"):
    """
    在地图上只显示给定 site_coords 范围内的区域，并保存为 PDF。
    """
    # 如果字典为空或只有一个站点，你可以手动设置范围，也可再改写逻辑
    if not site_coords:
        print("No site coordinates provided.")
        return

    # 把所有维度和经度单独取出
    lats = [coord[0] for coord in site_coords.values()]
    lons = [coord[1] for coord in site_coords.values()]

    # 找到最小和最大经纬度
    min_lat, max_lat = min(lats), max(lats)
    min_lon, max_lon = min(lons), max(lons)

    # 给地图范围加一点padding，防止点贴边太近
    lat_padding = (max_lat - min_lat) * 0.3 if (max_lat - min_lat) > 0 else 0.5
    lon_padding = (max_lon - min_lon) * 0.11 if (max_lon - min_lon) > 0 else 0.5

    # 设置地图显示的范围
    display_min_lat = min_lat - lat_padding
    display_max_lat = max_lat + lat_padding
    display_min_lon = min_lon - lon_padding
    display_max_lon = max_lon + lon_padding

    # 创建画布和投影
    fig = plt.figure(figsize=(8, 6))
    ax = fig.add_subplot(1, 1, 1, projection=ccrs.PlateCarree())

    # 设置地图显示的经纬度范围
    ax.set_extent([display_min_lon, display_max_lon,
                   display_min_lat, display_max_lat],
                  crs=ccrs.PlateCarree())

    # 添加一些地理特征
    ax.add_feature(cfeature.LAND, edgecolor='black', zorder=0)
    ax.add_feature(cfeature.OCEAN, zorder=0)
    ax.add_feature(cfeature.BORDERS, linestyle=':', zorder=1)
    ax.add_feature(cfeature.LAKES, alpha=0.5, zorder=1)
    ax.add_feature(cfeature.RIVERS, zorder=1)

    # 逐个站点打点与标注
    for site_name, (lat, lon) in site_coords.items():
        ax.plot(lon, lat,
                marker='o',
                color='red',
                markersize=5,
                transform=ccrs.PlateCarree())
        ax.text(lon + 0.1, lat + 0.1, site_name,
                transform=ccrs.PlateCarree(),
                fontsize=9,
                color='blue')

    # 保存为 PDF
    plt.savefig(output_pdf, format='pdf')
    plt.show()


# 调用函数，生成聚焦到所给站点的 PDF
plot_sites_zoom_in(site_coords, "fabric_sites_zoom.pdf")
```
